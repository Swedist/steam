```python
import requests
from fake_useragent import UserAgent

# Создаём сессию, чтобы cookies сохранялись между запросами
session = requests.Session()

# Настраиваем заголовки как у обычного браузера
session.headers.update({
    "User-Agent": UserAgent().random,  # случайный браузерный UA
    "Accept": "application/json, text/javascript, */*; q=0.01",
    "Accept-Language": "en-US,en;q=0.9",
    "Referer": "https://cs.money/",
    "X-Requested-With": "XMLHttpRequest",
    "Connection": "keep-alive",
})

# URL запроса к API CS.Money
url = (
    "https://cs.money/1.0/market/sell-orders?"
    "limit=60&offset=0&name=AK-47 | Emerald Pinstripe&"
    "quality=mw&isStatTrak=false&isSouvenir=false"
)

try:
    response = session.get(url, timeout=10)  # таймаут на случай зависаний
    response.raise_for_status()  # выбросит ошибку при статусе != 200
except requests.exceptions.HTTPError as errh:
    print("HTTP Error:", errh)
except requests.exceptions.ConnectionError as errc:
    print("Error Connecting:", errc)
except requests.exceptions.Timeout as errt:
    print("Timeout Error:", errt)
except requests.exceptions.RequestException as err:
    print("OOps: Something Else", err)
else:
    data = response.json()  # JSON-ответ
    print("Успешно получено элементов:", len(data))
    print(data[:3])  # первые 3 объекта для примера
```
